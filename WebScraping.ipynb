```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin

base_url = 'https://books.toscrape.com/'

response = requests.get(base_url)
soup = BeautifulSoup(response.text, 'html.parser')

category_links = soup.find('ul', class_='nav-list').find('ul').find_all('a')

categories = {}
for link in category_links:
    category_name = link.text.strip()
    category_url = urljoin(base_url, link['href'])
    categories[category_name] = category_url

print("URL-urile categoriilor:")
for name, url in categories.items():
    print(f"{name}: {url}")

all_category_pages = {}
for category, url in categories.items():
    page_num = 1
    category_pages = []
    
    while True:
        page_url = url.replace('index.html', f'page-{page_num}.html')
        response = requests.get(page_url)
        
        if response.status_code != 200:
            break
        
        category_pages.append(page_url)
        page_num += 1
    
    all_category_pages[category] = category_pages if category_pages else [url]

print("\nToate paginile din fiecare categorie:")
for category, pages in all_category_pages.items():
    print(f"{category}:")
    for page in pages:
        print(f"  {page}")
```

```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
url = 'https://books.toscrape.com/catalogue/category/books/travel_2/index.html'
def parse_books_from_page(url):
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')

    books = soup.find_all('article', class_='product_pod')
    two_star_books = []

    for book in books:
        rating = book.find('p')
        if rating:
            title = book.h3.a['title']  
            two_star_books.append(title)

    return two_star_books

all_two_star_books = []
all_two_star_books = parse_books_from_page(url)
print(all_two_star_books)
```
